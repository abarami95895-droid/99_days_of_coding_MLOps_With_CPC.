```python
sns.countplot(x='Age',hue='Alcohol consumption' ,data=data)
plt.title('Age vs Alcohol consumption')
plt.show()
```

```python
sns.countplot(x='Age',hue='Caffeine consumption' ,data=data)
plt.title('Age vs Caffeine consumption')
plt.show()
```
