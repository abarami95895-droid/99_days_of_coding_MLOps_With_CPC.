```python

data['Room'] = data['Room'].fillna(data['Room'].mean())
```

```python

data['Parking'] = data['Parking'].fillna(data['Parking'].mean())
```

```python

data['Warehouse'] = data['Warehouse'].fillna(data['Warehouse'].mean())
```

```python

data['Elevator'] = data['Elevator'].fillna(data['Elevator'].mean())
```

```python

data['Price'] = data['Price'].fillna(data['Price'].mean())
```

```python

data['Price(USD)'] = data['Price(USD)'].fillna(data['Price(USD)'].mean())
```
